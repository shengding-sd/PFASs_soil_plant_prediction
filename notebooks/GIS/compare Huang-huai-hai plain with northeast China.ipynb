```python
import pandas as pd
import rasterio
import numpy as np
```

```python
# read wheat
wheat = pd.read_csv('../../dataset/region prediction/wheat_clustered.csv')

# read TIFF
with rasterio.open('../../result/all/tif/long_wheat_root.tif') as src:
    coords = list(zip(wheat['lon'], wheat['lat']))
    values = [v[0] for v in src.sample(coords)]

wheat['Prediction'] = values

# remove NoData
wheat.loc[wheat['Prediction'] == src.nodata, 'Prediction'] = np.nan

# statistics by Cluster 
result = (
    wheat
    .groupby('Cluster')['Prediction']
    .agg(['mean', 'std', 'count'])
    .reset_index()
)

# Mean ± SD
result['Mean ± SD'] = (
    result['mean'].round(3).astype(str)
    + ' ± ' +
    result['std'].round(3).astype(str)
)

print(result)
```

```python
maize = pd.read_csv('../../dataset/region prediction/maize_clustered.csv')

with rasterio.open('../../result/all/tif/long_maize_root.tif') as src:

    coords = list(zip(maize['lon'], maize['lat']))
    values = [v[0] for v in src.sample(coords)]

maize['Prediction'] = values

maize.loc[maize['Prediction'] == src.nodata, 'Prediction'] = np.nan

result = (
    maize
    .groupby('Cluster')['Prediction']
    .agg(['mean', 'std', 'count'])
    .reset_index()
)

# Mean ± SD
result['Mean ± SD'] = (
    result['mean'].round(3).astype(str)
    + ' ± ' +
    result['std'].round(3).astype(str)
)

print(result)
```
